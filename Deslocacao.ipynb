```python
img = cv2.imread('pokemon.jpg')
img
```

```python
img_gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img
img_gray
```

#Deslocamento

```python
#Estou criando a deslocação no eixo x
dx = 50
dy = 0
altura = img_gray.shape[0]
largura = img_gray.shape[1]


transladada = np.zeros((altura,largura),dtype= np.uint8)#se ultrapassar para 255 volte para 0, isso é por causa no tipo de dado uint8
for x in range(largura):
  for y in range (altura):
    novo_x = x + dx
    novo_y = y + dy

    if 0 <= novo_x and novo_x < largura :
      if 0 <= novo_y and novo_y < altura :
        transladada[novo_y,novo_x] = img_gray[y,x]

plt.figure(figsize= (20,5))
plt.axis('off')
plt.title('Figura transladada')
plt.imshow(transladada, cmap='gray')
plt.show()
```

```python
#Estou criando a deslocação no eixo x
dx = 50
dy = 0
altura = img_gray.shape[0]
largura = img_gray.shape[1]


transladada = np.zeros((altura,largura),dtype= np.uint8)#se ultrapassar para 255 volte para 0, isso é por causa no tipo de dado uint8
for x in range(largura):
  for y in range (altura):
    novo_x = (x + dx) % largura
    novo_y = (y + dy) % altura

    transladada[novo_y,novo_x] = img_gray[y,x]

plt.figure(figsize= (20,5))
plt.title('Figura transladada')
plt.axis('off')
plt.imshow(transladada, cmap='gray')
plt.show()
```

#Escalonamento

```python
sX = 1.5
sY = 1.5
altura = img_gray.shape[0]
largura = img_gray.shape[1]

nova_largura = int(largura * sX)
nova_altura = int(altura * sY)

escalada = np.zeros((nova_altura,nova_largura),dtype= np.uint8)#se ultrapassar para 255 volte para 0, isso é por causa no tipo de dado uint8

for x in range(nova_largura):
  for y in range(nova_altura):

    novo_x = int(x / sX)
    novo_y = int( / sY)

    escalada[y,x]  = img_gray[novo_y,novo_x]


plt.figure(figsize= (20,5))
plt.imshow(escalada, cmap='gray')
plt.title('Figura escalonada')
plt.axis('off')
plt.show()
```

#Espelhamento

```python
altura = img_gray.shape[0]
largura = img_gray.shape[1]


Espelhada_h = np.zeros((nova_altura,nova_largura),dtype= np.uint8)#se ultrapassar para 255 volte para 0, isso é por causa no tipo de dado uint8

for x in range(largura):
  for y in range(altura):

    novo_x = largura -1 - x

    escalada[y,novo_x]  = img_gray[y,x]


plt.figure(figsize= (20,5))
plt.imshow(escalada, cmap='gray')
plt.title('Figura espelhada em X')
plt.axis('off')
plt.show()
```
