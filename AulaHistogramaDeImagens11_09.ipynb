```python
img = cv2.imread('garchomp.png')
img
```

```python
img_gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img_gray
```

```python
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
img_rgb
```

```python
def calc_hist(img):
  img_vetor = img.ravel() #pode usar flatten
  hist_vetor = np.zeros(256, dtype=np.int64)

  for px in img_vetor:
    hist_vetor[px] += 1

  return hist_vetor
```

```python
hist_img = calc_hist(img_gray)

plt.title('histograma de imagens em escala de cinza')
plt.bar(range(256),hist_img)
```
