```python
img = cv2.imread('teste.jpg')
print(img)
```

```python
cv2_imshow(img)
```

```python
img_mod = img/2
cv2_imshow(img_mod)
```

```python
img_mod = img * 2
cv2_imshow(img_mod)
```

```python
img2= img_mod
img_mod = np.maximum(img2,255)
cv2_imshow(img_mod)
```

```python
img_mod = img + 10
cv2_imshow(img_mod)
```

```python
#Aqui está dividindo a imagem por camadas, no caso são 3 camadas
B, G, R = cv2.split(img)
print('Com a biblioteca CV2')
cv2_imshow(R)
cv2_imshow(G)
cv2_imshow(B)

plt.figure(figsize=(10, 10))

#subplot do azul
plt.subplot(6,3,1)
plt.title( 'Vermelho(R)')
plt.imshow(R)
plt.axis('off') #Elimina os numeros do lado de cada plot
#subplot do verde
plt.subplot(6,3,2)
plt.title( 'Verde(G)')
plt.imshow(G)
plt.axis('off')
#subplot do azul
plt.subplot(6,3,3)
plt.title( 'Azul(B)')
plt.imshow(B)
plt.axis('off')

#Consigo ver qual das tres cores é predominante na imagem
plt.subplot(6,3,4)
plt.title( 'Vermelho(R)')
plt.imshow(R, cmap='gray')
plt.axis('off') #Elimina os numeros do lado de cada plot
plt.subplot(6,3,5)
plt.title( 'Verde(G)')
plt.imshow(G, cmap='gray')
plt.axis('off') #Elimina os numeros do lado de cada plot
plt.subplot(6,3,6)
plt.title( 'Azul(B)')
plt.imshow(B, cmap='gray')
plt.axis('off') #Elimina os numeros do lado de cada plot
```
