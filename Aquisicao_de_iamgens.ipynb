```python
img = cv2.imread('sapinho eletrico.png')
img
```

```python
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
img
img_rgb
```

```python
img_hsv = cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
img_hsv
```

```python
plt.figure(figsize=(20,15))
plt.subplot(2,1,1)
plt.imshow(img)
plt.title('Imagem BGR')
plt.axis('off')
plt.subplot(2,2,2)
plt.imshow(img_rgb)
plt.title('Imagem RGB')
plt.axis('off')
plt.show()
```

```python
fig,ax =plt.subplots(1,3,figsize=(10,7))
ax[0].imshow(img)
ax[0].set_title("Imagem BGR")
ax[0].axis('off')

ax[1].imshow(img_rgb)
ax[1].set_title('Imagem RGB')
ax[1].axis('off')

ax[2].imshow(img_hsv)
ax[2].set_title("Imagem HSV")
ax[2].axis('off')
```
